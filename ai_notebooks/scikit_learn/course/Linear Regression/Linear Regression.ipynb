```python
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
bmi_life_data = pd.read_csv("./data/bmi_and_life_expectancy.csv")
X = bmi_life_data[['BMI']]
y = bmi_life_data[['Life expectancy']]
```

```python
bmi_life_data
```

```python
bmi_life_model = LinearRegression()
bmi_life_model.fit(X, y)
```

```python
# Predict life expectancy for a BMI value of 21.07931
laos_life_exp = bmi_life_model.predict(pd.DataFrame([[21.07931]], columns=['BMI']))
laos_life_exp
```

```python
import matplotlib.pyplot as plt

plt.scatter(X, y, color='blue', alpha=0.6)
plt.plot(X, bmi_life_model.predict(X), color='red', linewidth=2)
plt.xlabel('BMI')
plt.ylabel('Life Expectancy')
plt.title('BMI vs Life Expectancy')
plt.show()
```
