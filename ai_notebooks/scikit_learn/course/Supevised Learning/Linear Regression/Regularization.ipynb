```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Lasso
```

```python
train_data = pd.read_csv('data/regularization_data.csv', header = None)
X = train_data.iloc[:,:-1]
y = train_data.iloc[:,-1]
```

```python
# Create the linear regression model with lasso regularization
lasso_reg = Lasso()

lasso_reg.fit(X, y)

# Retrieve and print out the coefficients from the regression model
reg_coef = lasso_reg.coef_
reg_coef
```

```python
lasso_reg.predict([[-3.90381,0.53243,2.83416,-5.42397,-0.06367,-0.22810]]) #6.05628
```

```python
import matplotlib.pyplot as plt

y_pred = lasso_reg.predict(X)

plt.scatter(y, y_pred, color='blue', alpha=0.6)
plt.plot([y.min(), y.max()], [y.min(), y.max()], color='red', linewidth=2)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Lasso Regression: Actual vs Predicted')
plt.show()
```
