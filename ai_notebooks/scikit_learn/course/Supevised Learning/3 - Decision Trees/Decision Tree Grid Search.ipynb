```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd
import numpy as np
```

```python
data = np.asarray(pd.read_csv('data/decision_tree_data.csv', header=None))
X = data[:,0:2]
y = data[:,2]
```

```python
from sklearn.model_selection import GridSearchCV

param_grid = {
    'max_depth': range(1, 20),
    'min_samples_split': range(2, 20),
    'min_samples_leaf': range(1, 20),
}

grid_search = GridSearchCV(
    DecisionTreeClassifier(),
    param_grid,
    scoring='accuracy',
    cv=5,
    refit=True,
)
grid_search.fit(X, y)

# Filter results with perfect accuracy and pick smallest params
results = pd.DataFrame(grid_search.cv_results_)
# perfect = results[results['mean_test_score'] == 1.0]
#
# if perfect.empty:
#     print("No combination achieved 1.0 accuracy")
# else:
#     # Sort by sum of params to find the smallest combination
#     perfect = perfect.sort_values(
#         by=['param_max_depth', 'param_min_samples_split', 'param_min_samples_leaf']
#     )
#     best = perfect.iloc[0]
#     print(f"max_depth:         {best['param_max_depth']}")
#     print(f"min_samples_split: {best['param_min_samples_split']}")
#     print(f"min_samples_leaf:  {best['param_min_samples_leaf']}")
#     print(f"accuracy:          {best['mean_test_score']}")

results
```
