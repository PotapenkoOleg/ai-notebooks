```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
import pandas as pd
import numpy as np
```

```python
data = np.asarray(pd.read_csv('data/decision_tree_data.csv', header=None))
X = data[:, 0:2]
y = data[:, 2]
```

```python
model = DecisionTreeClassifier(max_depth=6, min_samples_split=2, min_samples_leaf=3)
# max_depth=7, min_samples_split=2, min_samples_leaf=1
# max_depth=7, min_samples_split=2, min_samples_leaf=1
# max_depth=6, min_samples_split=3, min_samples_leaf=3
```

```python
model.fit(X, y)
```

```python
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt

plt.figure(figsize=(100, 50))
plot_tree(model, filled=True, rounded=True)
plt.show()
```

```python
y_pred = model.predict(X)
```

```python
y_pred
```

```python
acc = accuracy_score(y, y_pred)
```

```python
acc
```

```python
print(f"max_depth:         {model.get_depth()}")
print(f"min_samples_split: {model.min_samples_split}")
print(f"min_samples_leaf:  {model.min_samples_leaf}")
```
