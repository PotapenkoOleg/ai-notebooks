```python
import numpy as np
import pandas as pd
from sklearn.linear_model import Lasso
from sklearn.preprocessing import StandardScaler
```

```python
train_data = pd.read_csv('data/feature_scaling.csv', header = None)

X = train_data.iloc[:,:-1]
y = train_data.iloc[:,-1]
```

```python

scaler = StandardScaler()

# Fit the standardization parameters and scale the data.
X_scaled = scaler.fit_transform(X)

# Create the linear regression model with lasso regularization.
lasso_reg = Lasso()

lasso_reg.fit(X_scaled, y)

# Retrieve and print out the coefficients from the regression model.
reg_coef = lasso_reg.coef_
reg_coef
```

```python
import matplotlib.pyplot as plt

y_pred = lasso_reg.predict(X)

plt.scatter(y, y_pred, color='blue', alpha=0.6)
plt.plot([y.min(), y.max()], [y.min(), y.max()], color='red', linewidth=2)
plt.xlabel('Actual')
plt.ylabel('Predicted')
plt.title('Lasso Regression: Actual vs Predicted')
plt.show()
```
